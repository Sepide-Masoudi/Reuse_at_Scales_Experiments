```python
# First cell: dependencies

import matplotlib.pyplot as plt
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib import gridspec
from scipy import stats

warnings.filterwarnings("ignore")

plt.style.use("seaborn-v0_8-darkgrid")
sns.set_palette("husl")
```

```python
class ExperimentComparisonAnalyzer:
    def __init__(self, setups_dir, setup_names, metric_names, output_dir="./analysis_results"):
        self.setups_dir = setups_dir
        self.setup_names = setup_names
        self.metric_names = metric_names
        self.output_dir = output_dir

        self.all_data = {}
        self.summary = {}
        self.categories = {}

        os.makedirs(self.output_dir, exist_ok=True)

    # ------------------------------------------------------------------
    # Data loading
    # ------------------------------------------------------------------
    def load_metric_from_csv(self, setup_path, metric_name):
        csv_path = os.path.join(setup_path, f"{metric_name}.csv")
        if not os.path.exists(csv_path):
            return None
        try:
            return pd.read_csv(csv_path)
        except Exception as e:
            print(f"✗ Error loading {metric_name}.csv: {e}")
            return None

    def load_all_metrics(self):
        self.all_data = {}

        for setup in self.setup_names:
            setup_path = os.path.join(self.setups_dir, setup)
            if not os.path.exists(setup_path):
                print(f"Warning: {setup_path} not found")
                continue

            print(f"\n📂 Loading {setup}")
            setup_data = {}

            for metric in self.metric_names:
                df = self.load_metric_from_csv(setup_path, metric)
                if df is None or df.empty:
                    continue

                if "value" in df.columns:
                    values = df["value"]
                elif "Value" in df.columns:
                    values = df["Value"]
                else:
                    numeric_cols = df.select_dtypes(include=[np.number]).columns
                    if len(numeric_cols) == 0:
                        continue
                    values = df[numeric_cols[0]]

                values = values.dropna()
                if values.empty:
                    continue

                setup_data[metric] = {
                    "values": values.values,
                    "mean": values.mean(),
                    "std": values.std(),
                    "median": values.median(),
                    "min": values.min(),
                    "max": values.max(),
                    "q25": values.quantile(0.25),
                    "q75": values.quantile(0.75),
                    "sample_size": len(values),
                    "full_data": df,
                }

            self.all_data[setup] = setup_data

        return self.all_data

    # ------------------------------------------------------------------
    # Summary tables
    # ------------------------------------------------------------------
    def create_summary_dataframe(self):
        mean_df = pd.DataFrame(index=self.all_data.keys(), columns=self.metric_names)
        std_df = mean_df.copy()
        median_df = mean_df.copy()
        count_df = mean_df.copy()

        for setup, data in self.all_data.items():
            for metric in self.metric_names:
                if metric in data:
                    mean_df.loc[setup, metric] = data[metric]["mean"]
                    std_df.loc[setup, metric] = data[metric]["std"]
                    median_df.loc[setup, metric] = data[metric]["median"]
                    count_df.loc[setup, metric] = data[metric]["sample_size"]
                else:
                    count_df.loc[setup, metric] = 0

        self.summary = {
            "mean": mean_df.astype(float),
            "std": std_df.astype(float),
            "median": median_df.astype(float),
            "count": count_df.astype(int),
        }
        return self.summary

    # ------------------------------------------------------------------
    # Metric categorization
    # ------------------------------------------------------------------
    def categorize_metrics(self):
        cpu_energy, system, argo = [], [], []

        for m in self.metric_names:
            ml = m.lower()
            if any(k in ml for k in ["cpu", "power", "energy", "watts", "joules"]):
                cpu_energy.append(m)
            elif any(k in ml for k in ["argo", "workflow", "queue"]):
                argo.append(m)
            else:
                system.append(m)

        self.categories = {
            "CPU & Energy": cpu_energy,
            "System Performance": system,
            "Argo Workflows": argo,
        }
        return self.categories

    # ------------------------------------------------------------------
    # Plotting
    # ------------------------------------------------------------------
    def plot_metric_comparison(self, category_name, metrics):
        from math import isnan

        df = self.summary["mean"][metrics]
        fig, axes = plt.subplots(1, 3, figsize=(16, 6))

        # Grouped bars
        ax = axes[0]
        x = np.arange(len(df.index))
        width = 0.8 / len(df.columns)

        for i, col in enumerate(df.columns):
            ax.bar(x + i * width, df[col], width, label=col)

        ax.set_xticks(x + width)
        ax.set_xticklabels(df.index, rotation=45)
        ax.set_title(category_name)
        ax.legend(fontsize=8)
        ax.grid(True, alpha=0.3)

        # Heatmap
        ax = axes[1]
        norm = df / df.max()
        im = ax.imshow(norm.T, aspect="auto", cmap="RdYlBu_r")
        ax.set_yticks(range(len(df.columns)))
        ax.set_yticklabels(df.columns)
        ax.set_xticks(range(len(df.index)))
        ax.set_xticklabels(df.index, rotation=45)
        plt.colorbar(im, ax=ax)

        # Trends
        ax = axes[2]
        for col in df.columns:
            vals = df[col].values
            if not all(isnan(v) for v in vals):
                ax.plot(df.index, vals / np.nanmax(vals), "o-", label=col)
        ax.legend(fontsize=8)
        ax.grid(True)

        plt.tight_layout()
        plt.show()

    # ------------------------------------------------------------------
    # Dashboards and deep dives (unchanged logic)
    # ------------------------------------------------------------------
    def create_comprehensive_dashboard(self):
        create_comprehensive_dashboard(self.summary["mean"], self.all_data, self.categories)

    def detailed_metric_analysis(self, metric_name):
        create_detailed_metric_analysis(self.all_data, metric_name, self.setup_names)

    # ------------------------------------------------------------------
    # Export
    # ------------------------------------------------------------------
    def export_results(self):
        self.summary["mean"].to_csv(f"{self.output_dir}/mean_values.csv")
        self.summary["median"].to_csv(f"{self.output_dir}/median_values.csv")
        self.summary["std"].to_csv(f"{self.output_dir}/std_values.csv")
        self.summary["count"].to_csv(f"{self.output_dir}/sample_counts.csv")

        print(f"✓ Results exported to {self.output_dir}")

    # ------------------------------------------------------------------
    # One-shot runner
    # ------------------------------------------------------------------
    def run_full_analysis(self):
        print("📂 Loading data")
        self.load_all_metrics()

        print("📊 Creating summaries")
        self.create_summary_dataframe()

        print("📋 Categorizing metrics")
        self.categorize_metrics()

        for cat, metrics in self.categories.items():
            if metrics:
                self.plot_metric_comparison(cat, metrics)

        print("📈 Dashboard")
        self.create_comprehensive_dashboard()

        print("💾 Exporting")
        self.export_results()
```
